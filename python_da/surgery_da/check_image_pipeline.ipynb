```python
import nibabel as nib
from nilearn import plotting
import matplotlib.pyplot as plt
import os

DATA_PATH = "/Volumes/gabrielfriedman/Active/Friedman_Lab/PhysiologyBasalGanglia"

path = "sub-BG0003/ses-img01/anat/sub-BG0003_ses-img01_acq-mprage_ce-gad_run-01_T1w.nii.gz"

imaging_path = os.path.join(DATA_PATH, path)

print("File exists:", os.path.exists(imaging_path))

img = nib.load(imaging_path)

print("Shape:", img.shape)
print("Voxel dimensions (mm):", img.header.get_zooms()[:3])
print("Orientation:", nib.aff2axcodes(img.affine))

plotting.plot_anat(
    img,
    display_mode="ortho",
    title="MPRAGE MRI",
    draw_cross=True
)

plt.show()

print(img.affine)


print("qform code:", img.header["qform_code"])
print("sform code:", img.header["sform_code"])

print("qform matrix:\n", img.get_qform())
print("sform matrix:\n", img.get_sform())
```
